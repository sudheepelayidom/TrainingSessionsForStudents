**1: Library Import & Dataset Loading**

---



```python
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# Load dataset into pandas DataFrame
dataset1 = pd.read_csv("/content/diabetes.csv")
dataset1
```

**2: Target Variable Encoding**

```python
# Encoding categorical target column
dataset1["Outcome"] = dataset1["Outcome"].apply(lambda e: 0 if e in [0, "0", "N"] else 1)
dataset1
```

 **3: Duplicate Row Detection and Removal**

```python
# Check for duplicates across all columns
dataset1.duplicated()
# Remove duplicate rows directly in memory
dataset1.drop_duplicates(inplace=True)
dataset1
```

**4: Handle Missing Values**

```python
# BLOCK 4: FILL IN THE BLANKS (IMPUTATION)
from sklearn.impute import SimpleImputer

# 1. Fill missing 'Age' data using the Average (Mean Strategy)
imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
dataset1["Age"] = imputer.fit_transform(dataset1["Age"].values.reshape(-1, 1))

# 2. Fill missing 'Pregnancies' data using the Most Frequent number (Mode Strategy)
# We use mode here because you can't have 2.5 pregnancies! It must be a whole number.
imputer_mode = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
dataset1["Pregnancies"] = imputer_mode.fit_transform(dataset1["Pregnancies"].values.reshape(-1, 1))

print("Block 4 Complete: Missing values in 'Age' and 'Pregnancies' have been filled using statistics!")
```

**5: Outlier Detection and Filtering (IQR Method)**

```python
# BLOCK 5: DETECT AND REMOVE EXTREME OUTLIERS
# 1. Separate our 'Answers' (Outcome) from our 'Questions' (All other columns)
X = dataset1.drop(columns=['Outcome'])
y = dataset1['Outcome']

# 2. Calculate the statistical boundaries (Quartiles)
Q1 = X.quantile(0.25) # The 25% mark
Q3 = X.quantile(0.75) # The 75% mark
IQR = Q3 - Q1         # The distance between them

# 3. Create a rule to keep only the rows within normal, safe bounds
normal_mask = ~((X < (Q1 - 1.5 * IQR)) | (X > (Q3 + 1.5 * IQR))).any(axis=1)

# 4. Apply the rule to create our clean dataset
X_clean = X[normal_mask].copy()
y_clean = y[normal_mask].copy()

print(f"Block 5 Complete: Outliers removed!")
print(f"We safely threw away {X.shape[0] - X_clean.shape[0]} rows containing extreme mistakes.")
```

**6: TRAIN / TEST SPLIT**

```python

from sklearn.model_selection import train_test_split

# Split the data: 80% for learning (Train), 20% for testing (Test)
# random_state=42 ensures every student gets the exact same random shuffle
X_train, X_test, y_train, y_test = train_test_split(
    X_clean, y_clean, test_size=0.20, random_state=42, stratify=y_clean
)

print("Block 6 Complete: Data successfully split!")
print(f"The AI will learn from {X_train.shape[0]} patients (Training Set).")
print(f"The AI will be tested on {X_test.shape[0]} hidden patients (Testing Set).")
```

**7: Feature Scaling (Standardization)**

```python
#  NORMALIZE THE NUMBERS (SCALING)
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

# FIT (learn the math) and TRANSFORM (apply it) strictly on the Training Set
X_train_scaled = scaler.fit_transform(X_train)

# ONLY TRANSFORM the Test Set (Never 'fit' on the test set to prevent cheating!)
X_test_scaled = scaler.transform(X_test)

# Convert the results back into readable Pandas DataFrames
X_train_final = pd.DataFrame(X_train_scaled, columns=X_clean.columns)
X_test_final = pd.DataFrame(X_test_scaled, columns=X_clean.columns)

print("Block 7 Complete: All features are now scaled to a uniform range!")
display(X_train_final.head(3))
```

**8:Download the Preprocessed Dataset**

```python
# STITCH TOGETHER AND DOWNLOAD
from google.colab import files

# Reset the indexes so the features and target align perfectly
X_train_final.reset_index(drop=True, inplace=True)
y_train.reset_index(drop=True, inplace=True)

# Combine the clean features (X) and the clean answers (y) into one final table
final_clean_dataset = pd.concat([X_train_final, y_train], axis=1)

print("  Here is your final, machine-learning-ready dataset: ")
display(final_clean_dataset.head(5))

# Save to a CSV file
final_clean_dataset.to_csv("FINAL_preprocessed_diabetes_data.csv", index=False)

# Trigger the download
files.download("FINAL_preprocessed_diabetes_data.csv")
print("\nDownloading 'FINAL_preprocessed_diabetes_data.csv' to your computer...")
```
